### random forest
https://www.nvidia.cn/glossary/random-forest/
- 决策树对于新的样本并不太好
- 随机森林是监督式算法，由一组决策树组成。
- 输出是对一个问题最佳答案的共识！
- 用于分类或者回归。 回归的输出变量是一个数字序列；分类的输出是一个单一值。
- 每个决策树之间应该尽可能不相关。这通过**随机**实现

优缺点：
- 可以处理缺失值，因为有bagging和有放回抽样。
- 可以快速识别主要特征。
- 准确性由于决策树，但低于XGBOOST 梯度提升
- 包含大量树，速度比XGBOOST慢

工作原理：
- 通过一个“bagging”的过程，对训练集和特征进行抽样分组，最后汇总预测结果。
- 有放回抽样

#### Build first stump



Choose the min Gini index stump as the **first stump** in the forest

> Gini 指数（Gini Index）:衡量节点纯度。 值越小越纯
> $$Gini(t) = 1 - \sum_{i=1}^C p_i^2$$
> - C 是类别总数；
> - $ p_i $ 是样本中属于第 $ i $ 类的比例。
> 假设某个节点包含两个类别，正类和负类，正类占 70%，负类占 30%：
> $$Gini = 1 - (0.7^2 + 0.3^2) = 1 - (0.49 + 0.09) = 0.42$$
