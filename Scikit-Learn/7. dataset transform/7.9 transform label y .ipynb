### 7.9.2 label encoding
- 对类别标签进行编码，将字符串标签转换为整数标签。
- 用于二分类
- 类别的编码值随机，不应该用于反应任何信息
  - 如科学家 = 1， 程序员 = 4

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
y = ['cat', 'dog', 'fish', 'cat', 'dog']
y_encoded = le.fit_transform(y)
print(y_encoded) 
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
y = [1, 2, 3, 1, 2]
y_encoded = le.fit_transform(y)
print(y_encoded) 
```
