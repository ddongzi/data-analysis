# feature selection

我们应该如何筛选出关键特征。（不急于模型如lgbm）

Pearson相关：线性相关性。数值大小有意义，包括有序分类值

ANOVA (F检验) ：分类特征，类值无意义

## f_regression
不同类别下的平均target是否有区别？
- 如果区别大，说明分类特征有用

F检验统计量描述这个差异：组内方差和组间方差
$$
F = \frac{组间方差}{组内方差}
$$
比如有无空调对房价影响。
- 有空调：[3,3,6] 方差：6，均值4
-  无空调：[5,5,5] 方差：0，均值5
  
组间波动：$2*0.5^2 = 0.5$
F = 0.5 / 6

```python
from sklearn.datasets import make_regression
from sklearn.feature_selection import f_regression
X, y = make_regression(
    n_samples=50, n_features=3, n_informative=1, noise=1e-4, random_state=42
)
f_statistic, p_values = f_regression(X, y)
```

返回每个特征的f统计量值，f越大表示越显著越有用

p值0~1表示，显著检验的概率值. <0.05表示不是偶然发生的
