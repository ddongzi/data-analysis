# 新奇值和异常值检测

## 异常值检测

### isolation forest
随机切分特征，异常点远离样本，很快就能达到叶子节点。路径短

```python
import numpy as np
from sklearn.ensemble import  IsolationForest
import seaborn as sns

sns.set_theme(style='whitegrid', palette='coolwarm') 
rng = np.random.RandomState(0)
cov = np.array([[0.5,-0.1],[0.2, 0.3]])

n_samples, n_outliers = 100,10
```

```python
transformation = [[1, 1], [0.1, 0.5]] 
data_normal = np.dot(rng.randn(n_samples, 2), transformation)

data_outliers = rng.uniform(low=-4, high=4, size=(20, 2))
```

```python
X = np.concatenate([data_normal, data_outliers])
```

```python
clf = IsolationForest(n_estimators=100, contamination=0.05, random_state=42)
preds = clf.fit_predict(X)
```

```python
labels = ['Outlier' if p == -1 else 'Normal' for p in preds]
sns.scatterplot(
    x=X[:, 0], 
    y=X[:, 1], 
    hue=labels,          
    style=labels,     
)
```

predict：-1表示异常
